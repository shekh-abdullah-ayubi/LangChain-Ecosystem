```python
import os
from dotenv import load_dotenv
load_dotenv()
```

### Chat Models

**NVIDIA** base URL and API key directly into the ChatOpenAI class, as LangChain's OpenAI integration is completely compatible with OpenAI-compliant third-party endpoints.

```python
NV_API_KEY = os.getenv("NV_API_KEY")
NV_API_URL = os.getenv("NV_API_URL")
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

llm = ChatOpenAI(
    model="openai/gpt-oss-120b",
    openai_api_key=NV_API_KEY,
    openai_api_base=NV_API_URL,
    temperature=0.001,
    top_p=1,
    max_tokens=4096,
    streaming=False,
    extra_body={"reasoning_effort": "low"}
)

messages = [
    HumanMessage(content="What is Neo4j within 3 lines ?")
]

response = llm.invoke(messages)

print(response.content)
```

```python
llm.invoke("How to train ML models in brief?", extra_body={"reasoning_effort": "low"})
```
